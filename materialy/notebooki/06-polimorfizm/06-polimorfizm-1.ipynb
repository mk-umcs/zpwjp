# Polimorfizm

## Polimorfizm

Mechanizm polimorfizmu polega na tym, że tak samo nazwana operacja może oznaczać wykonanie zupełnie innej czynności w zależności od typów argumentów operacji. Np. operator `+` może w wielu językach (Python, Java) oznaczać zarówno dodawanie liczb (całkowite lub zmiennoprzecinkowe) jak i konkatenację (łączenie) napisów. W przypadku języków statycznie typowanych (np. Java) jest to przykład tzw. **polimorfizmu statycznego** &mdash; decyzja dotycząca wykonywanej operacji podejmowana jest w trakcie kompilacji.

W kontekście programowania obiektowego dużo częściej używa się terminu polimorfizm do opisania **polimorfizmu dynamicznego**. W tym przypadku decyzja dotycząca tego, która operacja ma zostać wykonana podejmowana jest w czasie wykonania programu. W języku Python (również w Javie) decyzja ta podjęta może być na podstawie typu obiektu, na rzecz którego wykonywana jest metoda. Np. w przypadku wywołania:

    obiekt.metoda(argument)

zostanie wywołana metoda z klasy, do której należy `obiekt`. Język Python jest dynamicznie typowany, więc typ `obiektu` może nie być znany do czasu wywołania metody.

Mechanizm ten dotyczy również operatorów (użycie operatorów w Pythonie oznacza wywołanie metod specjalnych). Np. to, co oznacza instrukcja

    a + b
    
zależy na pewno od typu obiektu `a` (wywołanie metody `__add__` z klasy obiektu `a`), a być może również od typu obiektu `b` (wywołanie metody `__radd__` z klasy obiektu `b`).

### Przykład &mdash; klasy definiowane przez użytkownika, metody

Mamy dwie klasy reprezentujące figury geometryczne płaskie. Klasa `Kolo` umożliwia reprezentowanie kół o zadanym promieniu. Klasa `Kwadrat` umożliwia reprezentowanie kwadratów o zadanym boku. Obie klasy mają metody `obwod` i `pole` zwracające odpowiednio obwód figury i jej pole powierzchni.

```
import math


class Kolo:
    def __init__(self, promien):
        self.promien = promien

    def obwod(self):
        return 2 * math.pi * self.promien

    def pole(self):
        return math.pi * self.promien**2
```

```
class Kwadrat:
    def __init__(self, bok):
        self.bok = bok

    def obwod(self):
        return 4 * self.bok

    def pole(self):
        return self.bok**2
```

Definiujemy funkcję `wyswietl_figure` używającą wspomnianych metod.

```
def wyswietl_figure(figura):
    print(f"obwód: {figura.obwod()}")
    print(f"pole: {figura.pole()}")
```

Wywołania metod `obwod` i `pole` są polimorficzne (używają mechanizmu **polimorfizmu**). Nie wiadomo której dokładnie metody dotyczą. Mogą oznaczać wywołanie metod z klasy `Kolo`:

```
wyswietl_figure(Kolo(3.5))
```

lub `Kwadrat`:

```
wyswietl_figure(Kwadrat(7))
```

Mogą to być również wywołania tych metod z jeszcze innej klasy z metodami `pole` i `obwod`.

W językach statycznie typowanych każdy parametr funkcji ma określony typ. Argument użyty przy wywołaniu musi być odpowiedniego typu. W przypadku funkcji takiej jak figura potrzebowalibyśmy np. typu `Figura`. To, że metoda może przyjmować argumenty różnych typów (dzięki czemu możliwy jest polimorfizm) jest w takich językach zapewniane przez różne mechanizmy takie jak np. interfejsy (niewystępujące w Pythonie jako konstrukcja językowa) i dziedziczenie (omówione dalej, ale nie jest wymagane do polimorfizmu w Pythonie). W Pythonie to, czy użycie obiektu zadanego typu jako argumentu danej funkcji się powiedzie (zakończy bez błędów) zależy wyłącznie od tego, czy można dla tego obiektu wywołać metody używane przez funkcję. Np. funkcję `wyswietl_figure` można wywołać z dowolnym obiektem mającym metody `pole` i `obwod` (czyli zachowującym się jak figura geometryczna). 

```
class A:
    def obwod(self):
        return 5

    def pole(self):
        return 8
```

```
a = A()
```

```
wyswietl_figure(a)
```

Liczba `5` nie ma metod `obwod` ani `pole`. Poniższe wywołanie zakończy się wyjątkiem [`AttributeError`](https://docs.python.org/3/library/exceptions.html#AttributeError) przy próbie odczytania pierwszej z nich:

```
wyswietl_figure(5)
```

```
class B:
    pass
```

```
b = B()
```

Podobnie obiekt klasy `B` nie ma wymaganych metod. Wywołanie `wyswietl_figure(b)` również spowoduje [`AttributeError`](https://docs.python.org/3/library/exceptions.html#AttributeError):

```
wyswietl_figure(b)
```

Zgodność typów określana na podstawie tego czy mają odpowiednie metody nazywana jest [Duck typing](https://en.wikipedia.org/wiki/Duck_typing) (&bdquo;kacze typowanie&rdquo;).

### Przykład &mdash; typy wbudowane, operatory

Użycie operatora `+` w poniższej funkcji również jest polimorficzne.

```
def suma(a, b):
    return a + b
```

Może on oznaczać dodawanie:

```
suma(4, 5)
```

Konkatenację:

```
suma("abc", "def")
```

Dowolną inną operację zdefiniowaną jako metoda `__add__` (lub `__radd__`) w odpowiedniej klasie (metody te są dokładniej omówione przy okazji omawiania metod specjalnych):

```
class A:
    def __add__(self, a):
        return A()

    def __repr__(self):
        return "A()"
```

```
suma(A(), A())
```
