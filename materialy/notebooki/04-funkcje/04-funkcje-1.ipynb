# Parametry, argumenty i wartości domyślne

**Parametr** (inaczej *parametr formalny*) &mdash; zmienna będąca częścią nagłówka funkcji. Np. poniższa funkcja `f` ma trzy parametry: `a`, `b` i `c`.

```
def f(a, b, c):
    return a * b + c
```

**Argument** (inaczej *parametr aktualny*) &mdash; wyrażenie przekazane podczas wywołania funkcji. Np. w poniższym wywołaniu argumentami są `x`, `3.4` i `y + 2.3`.

```
x = 3.2
y = -4.5
```

```
f(x, 3.4, y + 2.3)
```

Mówimy, że funkcja **ma** parametry i **przyjmuje** argumenty.

## Przekazywanie argumentów do funkcji w Pythonie

Przypomnijmy sposób przekazywania argumentów. Podczas wywołania parametr staje się lokalną nazwą wskazującą przekazany obiekt. Przypisanie do parametru zmienia lokalne powiązanie, a nie zmienną używaną przez wywołujący kod. Zmiana mutowalnego obiektu może natomiast być widoczna również poza funkcją.

```
def suma(a, b):
    a += b
    return a
```

```
x = 5
s = suma(x, 5)
```

```
s
```

```
x
```

Wywołanie funkcji `suma` nie zmieniło wartości zmiennej `x`. Zmodyfikowana została wyłącznie lokalna zmienna `a` (parametr), która przestała istnieć po zakończeniu działania funkcji.

## Parametry z wartością domyślną

Jeżeli parametr ma wartość domyślną, to możemy jej nie podawać przy wywołaniu funkcji.

```
def cena_brutto(cena_netto, vat=0.23):
    return cena_netto * (1 + vat)
```

Wywołanie z wartościami dla wszystkich parametrów:

```
cena_brutto(200, 0.08)
```

Brak wartości dla parametru `vat`. Użyta jest wartość domyślna, czyli `0.23`:

```
cena_brutto(100)
```

Parametry z wartością domyślną nazywane są często parametrami **opcjonalnymi** (ich użycie w wywołaniu funkcji jest opcjonalne) w przeciwieństwie do pozostałych **obowiązkowych** parametrów.

Kolejny przykład: obliczenie długości wektora. Obowiązkowa jedna współrzędna, dwie opcjonalne (obliczenia w jednym, dwóch lub trzech wymiarach zrealizowane jedną funkcją):

```
def dlugosc_wektora(x, y=0.0, z=0.0):
    return (x**2 + y**2 + z**2) ** 0.5
```

```
dlugosc_wektora(3.0)
```

```
dlugosc_wektora(3.0, 4.0)
```

```
dlugosc_wektora(3.0, 4.0, 12.0)
```

Błędy związane z nieprawidłowym przekazaniem argumentów do funkcji powodują zazwyczaj zgłoszenie wyjątku [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError). W poniższym wywołaniu nie podano obowiązkowej współrzędnej `x`, więc właśnie taki wyjątek zostanie zgłoszony:

```
dlugosc_wektora()
```

Parametry bez wartości domyślnej muszą poprzedzać te z wartością domyślną. Sama próba zdefiniowania poniższej funkcji, która nie przestrzega tej reguły, spowoduje wyjątek [`SyntaxError`](https://docs.python.org/3/library/exceptions.html#SyntaxError):

```
def cena_brutto(vat=0.23, cena_netto):
    return cena_netto * (1 + vat)
```

### Kiedy obliczana jest wartość parametru z wartością domyślną?

Wartości parametrów domyślnych obliczane są raz &mdash; podczas **definiowania** funkcji, a nie podczas wykonania. Wypróbuj poniższy przykład:

```
def wyswietl_imie_i_nazwisko(imie, nazwisko=input("nazwisko: ")):
    print(f"Witaj {imie} {nazwisko}!")
```

```
wyswietl_imie_i_nazwisko("Jan", "Nowak")
```

```
wyswietl_imie_i_nazwisko("Piotr")
```

```
wyswietl_imie_i_nazwisko("Jan")
```

Przykład z pytaniem o nazwisko jest dość sztuczny, ale dobrze pokazuje, kiedy obliczana jest wartość domyślna. Dalej zobaczymy bardziej naturalny przykład i błąd, który może wystąpić przy użyciu listy jako wartości domyślnej.

Jeżeli chcemy, żeby wartość brakującego parametru była obliczana podczas wywołania funkcji, to zazwyczaj robi się to tak:

```
def wyswietl_imie_i_nazwisko(imie, nazwisko=None):
    if nazwisko is None:
        nazwisko = input("nazwisko: ")
    print(f"Witaj {imie} {nazwisko}!")
```

```
wyswietl_imie_i_nazwisko("Jan", "Nowak")
```

```
wyswietl_imie_i_nazwisko("Piotr")
```

```
wyswietl_imie_i_nazwisko("Anna")
```

### Użycie list, słowników i innych modyfikowalnych obiektów jako domyślnych wartości parametrów

W związku z powyższym sposobem wyznaczania wartości parametrów z wartością domyślną trzeba bardzo uważać na obiekty mogące zmieniać swoją wartość (listy, słowniki, obiekty klas zdefiniowanych za pomocą konstrukcji `class`).

Przykład: funkcja dopisująca wynik (np. zawodów, eksperymentu) zadany pierwszym argumentem, do tabeli wyników zadanej drugim argumentem. Zwracana jest tabela z dopisanym wynikiem. Jeżeli tabela wyników nie została przekazana jako parametr, to zakładamy, że jest to pierwszy wynik i zwracana jest jednoelementowa lista.

```
def dopisz_wynik(wynik, tabela=[]):
    tabela.append(wynik)
    return tabela
```

Wywołanie z podaną tabelą wyników:

```
wyniki = [1, 2, 3]
```

```
dopisz_wynik(4, wyniki)
```

Wywołanie bez zadanej tabeli wyników:

```
dopisz_wynik(10)
```

Wynik jest prawidłowy, ale spróbujmy jeszcze raz:

```
dopisz_wynik(20)
```

Domyślna wartość parametru `tabela` jest obliczana tylko raz &mdash; jest nią pusta lista utworzona podczas definicji funkcji. Kolejne wywołania używające wartości domyślnej nie tworzą nowej listy, tylko używają tej już utworzonej. Modyfikacja tej listy (dodanie elementu przez funkcję `dopisz_wynik`) jest widoczna podczas kolejnych wywołań próbujących używać tej wartości domyślnej.

Jeżeli chcemy, żeby przy każdym wywołaniu z użyciem wartości domyślnej tworzona była nowa lista, to możemy funkcję zdefiniować podobnie do funkcji `wyswietl_imie_i_nazwisko`:

```
def dopisz_wynik(wynik, tabela=None):
    if tabela is None:
        tabela = []
    tabela.append(wynik)
    return tabela
```

```
dopisz_wynik(5, [1, 2, 3, 4])
```

```
dopisz_wynik(10)
```

```
dopisz_wynik(5)
```
