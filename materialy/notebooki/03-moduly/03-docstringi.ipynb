# Dokumentowanie modułów, klas i metod

Krótki napis na początku modułu, klasy albo funkcji jest **napisem dokumentacyjnym** (ang. *docstring*). Zazwyczaj zapisuje się je z użyciem potrójnego cudzysłowu. Opisuje przeznaczenie danego elementu i jest dostępny podczas działania programu jako atrubut `__doc__`. Bywa używany np. przez narzędzia pomocy takie jak funkcja [`help`](https://docs.python.org/3/builtins/functions.html#help).

W napisie dokumentacyjnym funkcji przydatne jest wyjaśnienie znaczenia parametrów i wyniku, gdy nie wynika ono jasno z ich nazw.  

- [Dokumentacja napisów dokumentacyjnych](https://docs.python.org/3/tutorial/controlflow.html#documentation-strings)

Przykład docstringów w module `trasy.py`:

```python
"""Klasy wspólne dla programów pracujących z trasami."""

import copy


class Wspolrzedne:
    """Współrzędne punktu: długość i szerokość geograficzna."""

    def __init__(self, dlugosc, szerokosc):
        self.dlugosc = dlugosc
        self.szerokosc = szerokosc

    def __repr__(self):
        return (
            f"Wspolrzedne({self.dlugosc!r}, {self.szerokosc!r})"
        )


class Trasa:
    """Nazwana trasa złożona z kolejnych punktów."""

    def __init__(self, nazwa):
        self.nazwa = nazwa
        self.przebieg = []

    def dodaj_punkt(self, punkt):
        """Dodaj punkt na końcu trasy."""
        self.przebieg.append(punkt)

    def __repr__(self):
        return (
            f"<Trasa nazwa={self.nazwa!r}, "
            f"przebieg={self.przebieg!r}>"
        )


class Uzytkownik:
    """Użytkownik z listą ulubionych tras."""

    def __init__(self, imie, nazwisko):
        self.imie = imie
        self.nazwisko = nazwisko
        self.ulubione_trasy = []

    def dodaj_ulubiona(self, trasa):
        """Dodaje ulubioną trasę."""
        self.ulubione_trasy.append(trasa)

    def __repr__(self):
        return (
            f"<Uzytkownik imie={self.imie!r}, "
            f"nazwisko={self.nazwisko!r}, "
            f"ulubione_trasy={self.ulubione_trasy!r}>"
        )

    def kopiuj_ulubione(self, inny):
        """Kopiuje ulubione trasy innego użytkownika."""
        self.ulubione_trasy.extend(
            copy.deepcopy(inny.ulubione_trasy)
        )

```

````
import trasy
````

Atrybut `__doc__` przechowuje treść napisu dokumentacyjnego:

````
trasy.__doc__
````

````
trasy.Trasa.__doc__
````

````
trasy.Trasa.dodaj_punkt.__doc__
````

Funkcja [`help`](https://docs.python.org/3/library/functions.html#help) łączy ten opis z informacjami o danym elemencie, np. nagłówkami metod:

````
help(trasy.Trasa.dodaj_punkt)
````

````
help(trasy.Trasa)
````

````
help(trasy)
````
