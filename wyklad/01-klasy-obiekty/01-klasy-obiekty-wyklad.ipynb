# Programowanie obiektowe &mdash; klasy, obiekty

W jaki sposób możemy reprezentować w programie dane złożone? Jeżeli chcemy np. przechowywać imię, nazwisko i rok urodzenia danej osoby, to możemy przechowywać je w trzech zmiennych:

```python
imie = "Jan"
nazwisko = "Kowalski"
rok_urodzenia = 2000
```

Dla każdej osoby, której dane chcemy przechowywać potrzebujemy wtedy trzech zmiennych. Jeżeli chcemy napisać funkcję dostającą informacje o osobie, to funkcja ta musi mieć trzy parametry:

```python
def wyswietl_wiek(imie, nazwisko, rok_urodzenia, rok):
    print(f"{imie} {nazwisko} ma {rok - rok_urodzenia} lat w roku {rok}.")
```

```python
wyswietl_wiek("Jan", "Kowalski", 2000, 2022)
```

Funkcja dostająca informacje o dwóch osobach potrzebuje już co najmniej sześciu parametrów:

```python
def wyswietl_porownanie_wieku(imie_1, nazwisko_1, rok_urodzenia_1, imie_2, nazwisko_2, rok_urodzenia_2):
    
    print(f"{imie_1} {nazwisko_1} ", end='')
    
    if rok_urodzenia_1 < rok_urodzenia_2:
        print("jest starszy(a) niż ", end='')
    elif rok_urodzenia_1 > rok_urodzenia_2:
        print("jest młodszy(a) niż ", end='')
    else:
        print("ma tyle samo lat co", end='')
    
    print(f"{imie_2} {nazwisko_2}.")
```

```python
wyswietl_porownanie_wieku("Jan", "Kowalski", 2000, "Anna", "Nowak", 1995)
```

Ten sposób reprezentacji danych nie wyraża w czytelny sposób tego, że trzy dane (imię, nazwisko i rok urodzenia) opisują ten sam obiekt (jedną osobę). Parametr `rok_urodzenia_1` funkcji `wyswietl_porownanie_wieku` nie jest powiązany z parametrem `nazwisko_1` bardziej niż np. z parametrem `nazwisko_2`.

Lepszym pomysłem będzie reprezentacja każdej osoby jako krotki:

```python
jan = ("Jan", "Kowalski", 2000)
```

```python
anna = ("Anna", "Nowak", 1995)
```

Wtedy powyższe funkcje mogą być zdefiniowane tak:

```python
def wyswietl_wiek(osoba, rok):
    print(f"{osoba[0]} {osoba[1]} ma {rok - osoba[2]} lat w roku {rok}.")
```

```python
def wyswietl_porownanie_wieku(osoba_1, osoba_2):

    print(f"{osoba_1[0]} {osoba_1[1]} ", end='')
    
    if osoba_1[2] < osoba_2[2]:
        print("jest starszy(a) niż ", end='')
    elif osoba_1[2] > osoba_2[2]:
        print("jest młodszy(a) niż ", end='')
    else:
        print("ma tyle samo lat co", end='')
    
    print(f"{osoba_2[0]} {osoba_2[1]}.")
```

Wywołania funkcji stały się czytelniejsze:

```python
wyswietl_wiek(jan, 2022)
```

```python
wyswietl_porownanie_wieku(jan, anna)
```

Niestety stało się to kosztem czytelności kodu funkcji. Nie jest w czytelny sposób wyrażone to, że `osoba[0]` to imię, a `osoba[1]` to nazwisko. Można funkcje napisać nieco czytelniej:

```python
def wyswietl_wiek(osoba, rok):
    imie, nazwisko, rok_urodzenia = osoba
    print(f"{imie} {nazwisko} ma {rok - rok_urodzenia} lat w roku {rok}.")
```

```python
def wyswietl_porownanie_wieku(osoba_1, osoba_2):

    imie_1, nazwisko_1, rok_urodzenia_1 = osoba_1
    imie_2, nazwisko_2, rok_urodzenia_2 = osoba_2
    
    print(f"{imie_1} {nazwisko_1} ", end='')
    
    if rok_urodzenia_1 < rok_urodzenia_2:
        print("jest starszy(a) niż ", end='')
    elif rok_urodzenia_1 > rok_urodzenia_2:
        print("jest młodszy(a) niż ", end='')
    else:
        print("ma tyle samo lat co", end='')
    
    print(f"{imie_2} {nazwisko_2}.")
```

```python
wyswietl_wiek(jan, 2022)
```

Jednak w dalszym ciągu musimy pamiętać, żeby krotka przekazana w argumencie funkcji przechowywała dane w odpowiedniej kolejności (takiej jak oczekuje funkcja).

Co będzie jeżeli przekażemy do funkcji krotki ze złą kolejnością danych?

```python
jan = (2000, "Kowalski", "Jan")
```

```python
anna = (1995, "Anna", "Kowalska")
```

Błędna kolejność nie zawsze spowoduje błąd. Czasem po prostu da bezsensowne wyniki.

```python
wyswietl_porownanie_wieku(jan, anna)
```

W poniższym przykładzie wystarczy spojrzeć na wyniki, żeby się zorientować w błędzie. Ale czasem błąd może nie być taki oczywisty:

```python
george_michael = ("Michael", "George", 1963)
elton_john = ("John", "Elton", 1947)

wyswietl_porownanie_wieku(george_michael, elton_john)
```

## Klasy

Lepszym rozwiązaniem będzie zdefiniowanie własnego typu specjalnie zaprojektowanego do konkretnego celu. W językach obiektowych taki typ nazywa się **klasą** (ang. *class*).

**Uwaga**: Ponieważ w Pythonie wszystkie typy (również wbudowane typy proste jak np. `int`) mają cechy, które w wielu innych językach obiektowych dotyczą wyłącznie takich własnych typów, termin &bdquo;klasa&rdquo; jest w często kontekście Pythona używany całkowicie zamiennie z terminem &bdquo;typ&rdquo;. Natomiast czasem spotyka się również materiały, w których stosowane jest rozróżnienie takie jak w innych językach: na typy (typy wbudowane) i klasy.

Do zdefiniowania nowej klasy służy konstrukcja `class`. Najprostsza możliwa definicja nowej klasy wygląda tak:

```python
class Osoba:
    pass
```

Taka definicja jest mało przydatna, ale wystarczy do pokazania podstawowych mechanizmów języka. Dalej definicja zostanie rozwinięta.

Standardową konwencją nazywania klasy jest użycie wielkiej litery na początku każdego wyrazu i małych liter poza tym. Np.
- `DaneOsobowe`,
- `WspolrzedneGeograficzne`.

**Dobra praktyka:** Standardem jest używanie w kodzie języka angielskiego (np. `PersonalData`, `GeographicCoordinates`) i dobrze się do tego przyzwyczajać ze względu na czytanie dokumentacji (oraz właśnie kodu). W materiałach celowo używany jest jednak język polski, aby ułatwić naukę i skupić się bezpośrednio na istotnych koncepcjach.

## Konstruktor (metoda  `__init__`) 

Zazwyczaj pola tworzy się od razu podczas tworzenia obiektu. Żeby to umożliwić klasę można zdefiniować w następujący sposób:

```python
class Osoba:
    
    def __init__(self, imie, nazwisko, rok_urodzenia):
        self.imie = imie
        self.nazwisko = nazwisko
        self.rok_urodzenia = rok_urodzenia
```

Funkcja `__init__` jest wywoływana automatycznie podczas tworzenia obiektu i jest nazywana **konstruktorem** (ang. *constructor*). Jest to przykład **metody** (ang. *method*). Metody są funkcjami powiązanymi z klasą i definiowanymi razem z nią.

**Uwaga:** Spotyka się czasem materiały mówiące, że metoda `__init__` nie jest konstruktorem, bo spełnia nieco inną rolę niż konstruktor w wielu innych obiektowych językach programowania. W Pythonie jest również metoda `__new__`, która razem w połączeniu z `__init__` pełnią funkcję konstruktora z tych &bdquo;innych języków&rdquo;. Konieczność jawnego pisania lub używania metody `__new__` jest rzadka, więc jej znajomość rzadko jest przydatna i nie jest tak istotna jak znajomość metody `__init__`. Poza tym w oficjalnej dokumentacji Pythona metoda `__init__` jest nazywana konstruktorem i taka terminologia jest  też używana w tych materiałach.

Poniższa instrukcja tworzy nowy obiekt i podstawia go pod zmienną:

```python
jan = Osoba("Jan", "Kowalski", 2000)
```

Instrukcja ta działa w następujący sposób:
* tworzony jest nowy obiekt typu `Osoba`,
* wywoływana jest funkcja `Osoba.__init__` &mdash; jako parametr `self` automatycznie przekazywany jest nowo-utworzony obiekt (nie ma go na liście argumentów wywołania), jako pozostałe parametry przekazywane są kolejne argumenty wywołania.
* nowoutworzony obiekt podstawiany jest pod zmienną `jan`.

Obiekt ma teraz utworzone w konstruktorze pola:

```python
jan.imie
```

```python
jan.rok_urodzenia
```

Przykład utworzenia kolejnego obiektu:

```python
anna = Osoba("Anna", "Nowak", 1950)
```

## Metody

Poniższa definicja klasy zawiera również definicję metody `wiek` zwracającej wiek osoby w zadanym roku. Tym razem jest to zwykła metoda a nie metoda specjalna:

```python
class Osoba:
    
    def __init__(self, imie, nazwisko, rok_urodzenia):
        self.imie = imie
        self.nazwisko = nazwisko
        self.rok_urodzenia = rok_urodzenia
        
    def wiek(self, rok):
        return rok - self.rok_urodzenia
```

## Parametr `self`

Pierwszy parametr (zgodnie z konwencją nazywany zawsze `self`) oznacza zawsze obiekt dla którego wywołujemy metodę. Nie umieszcza się go na liście argumentów wywołania. W wywołaniu jest przed kropką.

Dodajmy jeszcze metodę sprawdzającą czy osoba jest starsza od drugiej osoby:

```python
class Osoba:
    
    def __init__(self, imie, nazwisko, rok_urodzenia):
        self.imie = imie
        self.nazwisko = nazwisko
        self.rok_urodzenia = rok_urodzenia
        
    def wiek(self, rok):
        return rok - self.rok_urodzenia
    
    def starszy_od(self, druga_osoba):
        return self.rok_urodzenia < druga_osoba.rok_urodzenia
```

```python
jan = Osoba("Jan", "Kowalski", 2000)
```

```python
anna = Osoba("Anna", "Nowak", 1995)
```

W poniższym wywołaniu jako parametr `self` przekazana będzie zmienna `jan`. Jako parametr `druga_osoba` przekazana będzie zmienna `anna`.

```python
jan.starszy_od(anna)
```

Pola i metody określane są wspólną nazwą: **składowe** (ang. *members*) klasy.

## Metody `__repr__` i `__str__`

Oprócz metody `__init__` zazwyczaj definiuje się jeszcze metody specjalne odpowiadające za konwersję obiektów klasy na napis. Metodami takimi są `__repr__` i `__str__`.

Metoda `__str__` jest &bdquo;dla użytkownika programu&rdquo; tzn. powinna konwertować obiekt na napis zrozumiały i czytelny dla użytkownika. Np. jeżeli obiekt reprezentuje datę, to powinno być jasne jaka to jest data, a nie np. jak obiekt został utworzony w programie.

Metoda `__repr__` jest &bdquo;dla programisty&rdquo; tzn. służy do zamiany obiektu na napis, który umożliwia sprawdzenie wartości obiektu. Metoda `__repr__` jest używana np. przez funkcję wbudowaną `repr`. Jest też używana przez interpreter w trybie interaktywnym do wyświetlania wartości wyrażeń.

Zobaczmy jej działanie na przykładzie typu `Fraction` z modułu `fractions` (typ do operacji na ułamkach):

```python
from fractions import Fraction
```

```python
trzy_czwarte = Fraction(3, 4)
```

```python
repr(trzy_czwarte)
```

Metoda `__repr__` zostanie też użyta do wyświetlenia wartości w trybie interaktywnym:

```python
trzy_czwarte
```

Metoda `__str__` jest między innymi używana przez funkcje `str`, `print` i napisy formatujące:

```python
str(trzy_czwarte)
```

```python
print(trzy_czwarte)
```

```python
f"{trzy_czwarte}"
```

Zdefiniujmy odpowiednie metody `__repr__` i `__str__` w klasie `Osoba`:

```python
class Osoba:
    
    def __init__(self, imie, nazwisko, rok_urodzenia):
        self.imie = imie
        self.nazwisko = nazwisko
        self.rok_urodzenia = rok_urodzenia
        
    def wiek(self, rok):
        return rok - self.rok_urodzenia
    
    def starszy_od(self, druga_osoba):
        return self.rok_urodzenia < druga_osoba.rok_urodzenia
    
    def __repr__(self):
        return f"Osoba({self.imie!r}, {self.nazwisko!r}, {self.rok_urodzenia!r})"
    
    def __str__(self):
        return f"{self.imie} {self.nazwisko} urodzony w roku {self.rok_urodzenia}"
```

```python
jan = Osoba("Jan", "Kowalski", 2000)
```

```python
repr(jan)
```

```python
jan
```

```python
print(jan)
```

```python
str(jan)
```

Metoda `__repr__` powinna w miarę możliwości zwrócić napis w takiej postaci, że wpisanie go w trybie interaktywnym (lub w kodzie źródłowym programu) powinno spowodować utworzenie odpowiedniego obiektu.

Zapis `!r` użyty w napisie formatującym oznacza, że wyrażenie poprzedzające go jest przekształcane na napis za pomocą funkcji `repr` a nie `str`. Porównaj: 

```python
imie = "Jan"
```

```python
print(f"{imie}")
```

```python
print(f"{imie!r}")
```

```python
print(f"{jan}")
```

```python
print(f"{jan!r}")
```
