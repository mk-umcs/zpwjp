# Argumenty pozycyjne i nazwane

## Argumenty pozycyjne i nazwane
**Argumenty pozycyjne** (ang. *positional arguments*) &mdash; &bdquo;zwykłe&rdquo; przekazanie argumentu, to pod który parametr są podstawiane zależy od pozycji na liście argumentów w wywołaniu funkcji.

**Argumenty nazwane** (ang. *keyword arguments*) &mdash; podczas wywołania jawnie podane jest pod który parametr należy je podstawić.

```python
print(3, 4, 5, sep=", ", end="; ")
print(6, 7, 8, end=".", sep=", ")
```

Pierwsze trzy argumenty w obu powyższych wywołaniach są argumentami pozycyjnymi, a ostatnie dwa &mdash; nazwanymi.

Poniższe przykłady prezentują użycie argumentów pozycyjnych i nazwanych.

```python
def wyswietl_czas(godzina, minuta, sekunda):
    return f"{godzina:02}:{minuta:02}:{sekunda:02}"
```

Same argumenty pozycyjne:

```python
wyswietl_czas(3, 4, 55)
```

Godzina przekazana jest jako argument pozycyjny. Sekunda i minuta jako nazwane. Ich kolejność na liście argumentów nie ma znaczenia:

```python
wyswietl_czas(3, sekunda=30, minuta=20)
```

```python
wyswietl_czas(3, minuta=20, sekunda=30)
```

Same argumenty nazwane:

```python
wyswietl_czas(sekunda=55, godzina=12, minuta=15)
```

Argumenty pozycyjne muszą w wywołaniu występować przed nazwanymi. W poniższym zapisie po argumencie nazwanym `godzina=10` podano dwa argumenty pozycyjne, więc Python zgłosi wyjątek [`SyntaxError`](https://docs.python.org/3/library/exceptions.html#SyntaxError) jeszcze przed wywołaniem funkcji:

```python
wyswietl_czas(godzina=10, 10, 20)
```

### Argumenty nazwane i wartości domyślne

Mechanizm argumentów nazwanych dobrze współpracuje z parametrami z wartością domyślną. Zmieńmy teraz definicję funkcji `wyswietl_czas` (minuta i sekunda mają teraz domyślną wartość zero):

```python
def wyswietl_czas(godzina, minuta=0, sekunda=0):
    return f"{godzina:02}:{minuta:02}:{sekunda:02}"
```

```python
wyswietl_czas(12)
```

```python
wyswietl_czas(godzina=12)
```

```python
wyswietl_czas(6, sekunda=55)
```

```python
wyswietl_czas(8, 15)
```

Wartości jednego parametru nie można podać dwukrotnie, nawet jeżeli za każdym razem jest taka sama. Tutaj wartość parametru `godzina` przekazujemy raz jako pierwszy argument pozycyjny, a drugi raz jako argument nazwany. Takie wywołanie spowoduje wyjątek [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

```python
wyswietl_czas(5, minuta=3, godzina=5)
```

Wartość każdego parametru musi być podana przez argument pozycyjny, argument nazwany albo wartość domyślną. Parametr `godzina` nie ma wartości domyślnej, a w poniższym wywołaniu nie przekazano jego wartości. Z tego powodu otrzymamy wyjątek [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

```python
wyswietl_czas(minuta=3, sekunda=6)
```

### Wymuszenie przekazywania argumentów jako pozycyjnych lub jako nazwanych

Przy porównywaniu sposobów wywołania użyjemy ponownie funkcji formatującej czas:

```python
def wyswietl_czas(godzina, minuta=0, sekunda=0):
    return f"{godzina:02}:{minuta:02}:{sekunda:02}"
```

Przekazywanie argumentów jako argumentów nazwanych może poprawić czytelność wywołań funkcji. Zobaczmy to na przykładzie funkcji [`datetime`](https://docs.python.org/3/library/datetime.html) z modułu [`datetime`](https://docs.python.org/3/library/datetime.html):

```python
import datetime
```

W poniższym wywołaniu ciężko zorientować się, który parametr za co odpowiada:

```python
dt = datetime.datetime(1, 2, 3, 4, 5, 6)
```

```python
dt.minute
```

```python
dt.month
```

Poniższe wywołanie jest czytelniejsze (ale niestety moduł [`datetime`](https://docs.python.org/3/library/datetime.html) nie wymusza tego):

```python
dt = datetime.datetime(year=1, month=2, day=3, hour=4, minute=5, second=6)
```

Używając wyłącznie argumentów pozycyjnych łatwo możemy omyłkowo przekazać parametry w złej kolejności:

```python
wyswietl_czas(s, m, h)
```

#### Wymuszenie przekazania argumentów jako nazwanych.

Funkcja może wymuszać wyłącznie argumenty nazwane dla wybranych parametrów. Np. funkcja [`sorted`](https://docs.python.org/3/library/functions.html#sorted) ma parametr `reverse` określający, czy kolejność sortowania ma być odwrócona:

```python
help(sorted)
```

```python
sorted([3, 4, 5, 2, 0, 1, 2], reverse=True)
```

Funkcja [`sorted`](https://docs.python.org/3/library/functions.html#sorted) przyjmuje tylko jeden argument pozycyjny. Próba przekazania wartości parametru `reverse` jako drugiego argumentu pozycyjnego zakończy się wyjątkiem [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

```python
sorted([3, 4, 5, 2, 0, 1, 2], True)
```

Komunikat o błędzie oznacza, że oczekiwany był tylko jeden argument pozycyjny. Argument `reverse` musi być przekazany jako nazwany.

Wymuszenie takie realizuje się przez umieszczenie znaku `*` na liście parametrów.
Wszystkie parametry po nim muszą być przekazane jako nazwane.

Poniższy przykład pokazuje kiedy może to być szczególnie przydatne. Zdefiniujmy najpierw funkcję bez użycia takiego wymuszenia.

```python
def naglowek_wiadomosci(adresat, temat, pilne=False):
    if pilne:
        temat = f"!!PILNE!! {temat}"
    return f"Do: {adresat}\nTemat: {temat}\n"
```

```python
print(naglowek_wiadomosci("Anna", "Zebranie"))
```

```python
print(naglowek_wiadomosci("Anna", "Zebranie", True))
```

Co będzie jeżeli wiadomość będzie pilna, ale zapomnimy o temacie w wywołaniu funkcji:

```python
print(naglowek_wiadomosci("Anna", True))
```

Użyliśmy omyłkowo `True` jako wartości parametru `temat`. Nie została podana wartość parametru `pilne`, ale ma on wartość domyślną.

Poprawmy definicję. Teraz parametry `temat` i `pilne` muszą być przekazane jako nazwane:

```python
def naglowek_wiadomosci(adresat, *, temat, pilne=False):
    if pilne:
        temat = f"!!PILNE!! {temat}"
    return f"Do: {adresat}\nTemat: {temat}\n"
```

Po zmianie definicji tylko adresat może być przekazany jako argument pozycyjny. Powtórzenie wcześniejszego problematycznego wywołania spowoduje więc wyjątek [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError): drugi argument pozycyjny nie może już zostać przyjęty jako wartość parametru `temat`.

```python
print(naglowek_wiadomosci("Anna", True))
```

Poprawne wywołanie z jawnym tematem i oznaczeniem pilności:

```python
print(naglowek_wiadomosci("Anna", temat="spotkanie", pilne=True))
```

Kolejny przykład: funkcja `wyswietl_czas`, ale z wymuszeniem wszystkich argumentów nazwanych:

```python
def wyswietl_czas(*, godzina, minuta, sekunda):
    return f"{godzina:02}:{minuta:02}:{sekunda:02}"
```

Wywołania są teraz czytelniejsze (wiadomo, który parametr za co odpowiada):

```python
wyswietl_czas(godzina=10, minuta=5, sekunda=55)
```

Trudniej o błąd z zamianą kolejności parametrów:

```python
h, m, s = 10, 5, 55
```

Ta wersja funkcji `wyswietl_czas` wymaga, żeby wszystkie argumenty były nazwane. Wywołanie jej z trzema argumentami pozycyjnymi zostanie odrzucone z wyjątkiem [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError), zamiast niepostrzeżenie zamienić godzinę z sekundą:

```python
wyswietl_czas(s, m, h)
```

Nie można teraz użyć argumentów pozycyjnych. Powyższe wywołanie musiałoby wyglądać tak (błąd jest tutaj od razu widoczny):

```python
wyswietl_czas(godzina=s, minuta=m, sekunda=h)
```

#### Wymuszenie przekazania argumentów jako pozycyjnych

Czasem nazwa danego parametru nie jest istotna i nie chcemy, żeby była widoczna w wywołaniach funkcji. Jeżeli nie jest istotna, to mogła np. nie być dobrze przemyślana i chcielibyśmy ją zmienić w definicji funkcji. Jeżeli nazwa ta mogła być użyta w wywołaniach z argumentem nazwanym, to po jej zmianie musimy ją również zmienić we wszystkich takich wywołaniach.

Jeżeli chcemy, żeby dało się użyć wyłącznie argumentu pozycyjnego dla danego parametru, to musimy go umieścić przed znakiem `/` na liście parametrów.

Przykład: obliczenie długości przedziału. Zadane są końce przedziału w dowolnej kolejności. Nazwy końców przedziałów nie są istotne:

```python
def dlugosc_przedzialu(a, b, /):
    return abs(a - b)
```

```python
dlugosc_przedzialu(4.5, 3)
```

Parametry `a` i `b` poprzedzają separator `/`, więc ich wartości trzeba podać pozycyjnie. Użycie ich nazw w poniższym wywołaniu spowoduje wyjątek [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

```python
dlugosc_przedzialu(a=4.5, b=3)
```

Kolejny przykład: funkcja obliczająca [średnią ważoną](https://pl.wikipedia.org/wiki/%C5%9Arednia_wa%C5%BCona).

Parametry:
- `dane` &mdash; sekwencja liczb (wyłącznie argument pozycyjny),
- `wagi` &mdash; wagi do średniej ważonej (argument pozycyjny lub nazwany),
- `dokladnosc` &mdash; jeżeli jest podany, to wynik jest zaokrąglany (wyłącznie argument nazwany).

Przykład średniej zakłada niepuste dane, taką samą liczbę wag jak danych i niezerową sumę wag. Skupiamy się na rodzajach parametrów, nie na walidacji danych wejściowych.

```python
def srednia(dane, /, wagi=None, *, dokladnosc=None):
    if wagi is None:
        wagi = [1] * len(dane)
    wynik = sum(d * w for d, w in zip(dane, wagi)) / sum(wagi)
    if dokladnosc is not None:
        wynik = round(wynik, dokladnosc)
    return wynik
```

Przykłady poprawnego użycia:

```python
srednia([1, 2])
```

```python
srednia([1, 2], [1, 7])
```

```python
srednia([1, 2, 2, 2, 2, 2, 2], dokladnosc=4)
```

```python
srednia([1, 2], [1, 7], dokladnosc=1)
```

```python
srednia([1, 2], wagi=[1, 7])
```

```python
srednia([1, 2], wagi=[1, 7], dokladnosc=1)
```

```python
srednia([1, 2], dokladnosc=1, wagi=[1, 7])
```

Parametr `dane` musi otrzymać wartość jako argument pozycyjny. Próba przekazania jej pod nazwą `dane` zakończy się wyjątkiem [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

```python
srednia(dane=[1, 2])
```

Wartość parametru `dokladnosc` można przekazać wyłącznie jako argument nazwany. W poniższym wywołaniu podano ją jako trzeci argument pozycyjny, którego funkcja nie przyjmuje; otrzymamy wyjątek [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

```python
srednia([1, 2], [1, 7], 1)
```
