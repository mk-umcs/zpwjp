# Zmienna liczba argumentów i dowolne argumenty nazwane

## Zmienna liczba argumentów pozycyjnych

Niektóre funkcje mogą przyjąć dowolną liczbę argumentów pozycyjnych. Przykładem takiej funkcji jest [`print`](https://docs.python.org/3/library/functions.html#print):

```python
print()
```

```python
print(1, 2, 3)
```

```python
print("Stoi", "na", "stacji", "lokomotywa")
```

Parametr poprzedzony gwiazdką w definicji funkcji &bdquo;zbiera&rdquo; wszystkie końcowe argumenty pozycyjne jako krotkę. Dla poniższej definicji wszystkie argumenty poza pierwszym będą dostępne jako krotka `xs`.

```python
def srednia_arytmetyczna(x, *xs):
    print(f"x = {x}, xs = {xs}")
    return (x + sum(xs)) / (1 + len(xs))
```

```python
srednia_arytmetyczna(1.0, 2.0, 3.0, 4.0)
```

```python
srednia_arytmetyczna(1.0)
```

Parametr `x` wymaga przynajmniej jednego argumentu pozycyjnego. Wywołanie bez argumentów nie poda jego wartości i spowoduje wyjątek [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

```python
srednia_arytmetyczna()
```

Wszystkie argumenty, które na liście parametrów są za parametrem z gwiazdką muszą być przekazane jako nazwane (inaczej byłyby też &bdquo;wyłapane&rdquo; przez parametr z gwiazdką), gwiazdka działa tu też tak jak sama gwiazdka na liście parametrów.

Wypróbujmy zmodyfikowaną definicję funkcji `srednia_arytmetyczna` z opcjonalnym wyświetlaniem wartości parametrów:

```python
def srednia_arytmetyczna(x, *xs, wyswietl=False):
    if wyswietl:
        print(f"x = {x}, xs = {xs}")
    return (x + sum(xs)) / (1 + len(xs))
```

Wartość parametru `wyswietl` musi być przekazana jako argument nazwany. Poniższy przykład potraktuje argument `True` jako wartość czwartą wartość liczbową równą 1.

```python
srednia_arytmetyczna(10, 30, 40, True)
```

Jedyny sposób, to:

```python
srednia_arytmetyczna(10, 30, 40, wyswietl=True)
```

Parametr `*xs` już wyznacza granicę, po której wartości parametrów trzeba przekazywać jako argumenty nazwane. Nie dodajemy za nim osobnego separatora `*`: poniższa próba zdefiniowania funkcji z obiema konstrukcjami jest niepoprawna składniowo i zgłosi wyjątek [`SyntaxError`](https://docs.python.org/3/library/exceptions.html#SyntaxError):

```python
def srednia_arytmetyczna(x, *xs, *, wyswietl=False):
    if wyswietl:
        print(f"x = {x}, xs = {xs}")
    return (x + sum(xs)) / (1 + len(xs))
```

## Rozwijanie sekwencji do listy argumentów

Elementy sekwencji można w łatwy sposób przekazać jako kolejne argumenty wywołania funkcji. Gwiazdka użyta przed argumentem będącym sekwencją &bdquo;rozbije&rdquo; tę sekwencję na kolejne parametry.

Mamy funkcję, która zwraca sumę swoich trzech parametrów:

```python
def suma_3(a, b, c):
    return a + b + c
```

Mamy sekwencję (np. listę) z trzema elementami do zsumowania:

```python
skladniki = [1, 2, 3]
```

Używając powyższej funkcji możemy to zrobić np. tak:

```python
suma_3(skladniki[0], skladniki[1], skladniki[2])
```

Funkcja `suma_3` wymaga trzech argumentów. Poniżej przekazujemy tylko jeden — listę z trzema elementami. Python nie przypisze automatycznie jej elementów do trzech parametrów, więc brak wartości dla dwóch z nich spowoduje wyjątek [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

```python
suma_3(skladniki)
```

Możemy jednak zrobić to tak jak poniżej. Wszystkie elementy sekwencji zostaną użyte jako kolejne argumenty pozycyjne funkcji:

```python
suma_3(*skladniki)
```

```python
skladniki.append(4)
```

Po dopisaniu elementu lista `skladniki` ma już cztery elementy. Gwiazdka przekaże je jako cztery osobne argumenty pozycyjne, ale funkcja `suma_3` przyjmuje tylko trzy. To wywołanie zakończy się wyjątkiem [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

```python
suma_3(*skladniki)
```

Szczególnie przydatne jest to, gdy chcemy użyć funkcji, która przyjmuje dowolną liczbę argumentów, a elementy są przechowywane w sekwencji, ale nie jest znana jej długość, lub sekwencja jest długa. Jeżeli długość nie jest znana, to nie możemy przekazać wszystkich argumentów osobno. Jeżeli sekwencja jest długa, to jest to niepraktyczne.

Przykład:

Powtarzamy potrzebną definicję średniej, aby ten przykład można było wykonać bez szukania jej w poprzedniej części:

```python
def srednia_arytmetyczna(x, *xs, wyswietl=False):
    if wyswietl:
        print(f"x = {x}, xs = {xs}")
    return (x + sum(xs)) / (1 + len(xs))
```

```python
xs = [i**2 for i in range(1000)]
```

Teoretycznie da się użyć wywołania `srednia_arytmetyczna(xs[0], ... xs[999])`, ale byłoby to wywołanie z tysiącem argumentów. Jest to kompletnie niepraktyczne i oczywiście wygodniej użyć poniższego wywołania:

```python
srednia_arytmetyczna(*xs, wyswietl=True)
```

Kolejne przykłady użycia gwiazdki przed argumentem wywołania:

```python
srednia_arytmetyczna(0, *skladniki, 4, wyswietl=True)
```

```python
srednia_arytmetyczna(*range(5, 8), wyswietl=True)
```

```python
srednia_arytmetyczna(*range(10), 10, *skladniki, 11, wyswietl=True)
```

### (*) Przykład z funkcją [`zip`](https://docs.python.org/3/library/functions.html#zip) (i brakiem funkcji `unzip`):

Funkcja [`zip`](https://docs.python.org/3/library/functions.html#zip) pozwala połączyć osobne sekwencje w pojedynczą sekwencję krotek. Pierwsze elementy wszystkich sekwencji utworzą pierwszą krotkę, drugie &mdash; drugą, itd.

```python
numery = [1, 2, 3, 4]
nazwy = ["Świdnik", "Lublin", "Warszawa", "Tokio"]
populacja = [40_000, 400_000, 1_600_000, 10_000_000]
```

```python
miasta = list(zip(numery, nazwy, populacja))
```

```python
miasta
```

W bibliotece standardowej nie ma funkcji odwrotnej, czyli funkcji, która rozbija listę takich krotek np. na krotkę osobnych sekwencji.

Ale zastanówmy się jaki wynik da takie wywołanie:

```python
list(
    zip(
        (1, "Świdnik", 40000),
        (2, "Lublin", 400000),
        (3, "Warszawa", 1600000),
        (4, "Tokio", 10000000),
    )
)
```

Przy zamianie sekwencji numerów, nazw i populacji na sekwencję miast (krotek) wywołanie [`zip`](https://docs.python.org/3/library/functions.html#zip) ma zawsze trzy argumenty: numery, nazwy i populacje. Przy zamianie w odwrotną stronę &mdash; miast na trzy osobne sekwencje &mdash; liczba argumentów funkcji [`zip`](https://docs.python.org/3/library/functions.html#zip) to liczba miast. Jeżeli więc mamy miasta w liście, to musimy właśnie użyć argumentu z gwiazdką:

```python
list(zip(*miasta))
```

Nie jest to do końca operacja odwrotna &mdash; w wyniku mamy listę krotek, a początkowo dane były w listach, ale w razie potrzeby możemy sobie te krotki łatwo skonwertować na listy.

## Funkcje przyjmujące dowolne argumenty nazwane

W podobny sposób jak możemy napisać funkcję, która przyjmuje dowolną liczbę parametrów pozycyjnych używając gwiazdki przed jednym z parametrów, tak samo możemy napisać funkcję, która przyjmie dowolne argumenty nazwane używając dwóch gwiazdek przed parametrem. Wartością tego parametru będzie wtedy słownik. Kluczami w tym słowniku będą nazwy argumentów, a wartościami &mdash; wartości tych argumentów.

Przykład:
Funkcja `filtruj` filtruje listę słowników według podanych kryteriów. Zwraca tylko te elementy, w których pola zadane argumentami nazwanymi mają zadaną wartość.

```python
def filtruj(dane, **filtry):
    print(f"filtry: {filtry}")
    pasujace = []
    for element in dane:
        pasuje = True
        for nazwa, wartosc in filtry.items():
            if element[nazwa] != wartosc:
                pasuje = False
                break
        if pasuje:
            pasujace.append(element)
    return pasujace
```

Przykłady użycia:

```python
produkty = [
    {"nazwa": "krzesło", "typ": "mebel", "cena": 300},
    {"nazwa": "biurko", "typ": "mebel", "cena": 2000},
    {"nazwa": "rower", "typ": "sprzęt sportowy", "cena": 2000},
    {"nazwa": "piłka", "typ": "sprzęt sportowy", "cena": 200},
]

uczniowie = [
    {"imie": "Jan", "nazwisko": "Kowalski", "ocena": 5},
    {"imie": "Anna", "nazwisko": "Nowak", "ocena": 3},
    {"imie": "Piotr", "nazwisko": "Kowalski", "ocena": 2},
    {"imie": "Joanna", "nazwisko": "Wójtowicz", "ocena": 5},
    {"imie": "Piotr", "nazwisko": "Wójtowicz", "ocena": 3},
    {"imie": "Piotr", "nazwisko": "Nowak", "ocena": 3},
    {"imie": "Beata", "nazwisko": "Wójcik", "ocena": 4},
]
```

```python
filtruj(produkty, typ="mebel")
```

```python
filtruj(produkty, typ="mebel", cena=2000)
```

```python
filtruj(uczniowie, imie="Piotr")
```

```python
filtruj(uczniowie, imie="Piotr", ocena=3)
```

```python
filtruj(uczniowie, ocena=6)
```

```python
filtruj(produkty)
```

Nie możemy tutaj użyć w definicji konkretnych parametrów dla każdego filtra, bo ani nazwy filtrów ani ich liczba nie są w żaden sposób ograniczone.

Podobny mechanizm przyjmowania dowolnych argumentów nazwanych wykorzystuje metoda [`dict.update`](https://docs.python.org/3/library/stdtypes.html#dict.update). Możemy przekazać jej słownik z nowymi wpisami albo podać wpisy jako argumenty nazwane. Nazwy argumentów stają się kluczami, a podane wartości — wartościami wpisów:

```python
d = {"a": 1, "b": 2}
```

```python
d.update({"b": 3, "c": 4})
```

```python
d.update(c=10, d=11)
```

```python
d
```

## Rozwijanie słowników do argumentów nazwanych

W taki sam sposób jak wartości parametrów pozycyjnych można wziąć z sekwencji poprzedzonej gwiazdką, tak wartości parametrów nazwanych można wziąć ze słownika (lub innego tzw. [**mapującego** typu danych](https://docs.python.org/3/glossary.html#term-mapping), ang. *mapping*) poprzedzonego dwiema gwiazdkami. Na przykład parametry `sep` i `end` funkcji [`print`](https://docs.python.org/3/library/functions.html#print) możemy przekazać w następujący sposób:

```python
parametry = {"sep": " + ", "end": ";"}
print(1, 2, 3, **parametry)
```

Jest to równoważne wywołaniu:

```python
print(1, 2, 3, sep=" + ", end=";")
```

W powyższym przykładzie nie widać korzyści z użycia tego mechanizmu. Wersja z jawnym użyciem argumentów nazwanych jest prostsza. Rozważmy teraz bardziej praktyczny przykład użycia.

Załóżmy, że nasz program przechowuje pewne dane w słowniku. Chcemy napisać funkcję zapisującą te dane do pliku w formacie [JSON](https://www.json.org/). Chcemy umożliwić dokładną konfigurację zapisu danych za pomocą parametrów nazwanych:
- czy wynikowy plik ma używać do wcięć spacji zamiast tabulatorów (parametr logiczny `spacje`, domyślnie `True`),
- jeżeli używamy spacji, to ilu spacji użyć zamiast jednego tabulatora (parametr `rozmiar_tab`, domyślnie `4`),
- jaka jest nazwa pliku (parametr `nazwa_pliku`, domyślnie `wyniki.json`),
- czy klucze słownika mają być w wynikowym pliku posortowane (parametr `sortuj_klucze`, domyślnie `True`),
- czy plik wynikowy ma używać zestawu znaków ograniczonego do ASCII, czyli np. napis `współdzielony` zostanie zapisany jako `wsp\u00f3\u0142dzielony` (parametr logiczny `ascii`, domyślnie `False`).

Funkcja używa modułu [`json`](https://docs.python.org/3/library/json.html) do konwersji skonstruowania danych w formacie JSON.

```python
import json


def zapisz_wyniki(
    dane,
    spacje=True,
    rozmiar_tab=4,
    nazwa_pliku="wyniki.json",
    sortuj_klucze=True,
    ascii=False,
):
    if spacje:
        indentacja = rozmiar_tab
    else:
        indentacja = "\t"
    with open(nazwa_pliku, "w", encoding="utf-8") as p:
        json.dump(
            dane, p, ensure_ascii=ascii, indent=indentacja, sort_keys=sortuj_klucze
        )
```

Przykład użycia funkcji:

```python
dane = {"id": 123, "nazwisko": "Kowalski", "imię": "Jan", "oceny": [3, 3, 2, 5]}
```

```python
zapisz_wyniki(dane, rozmiar_tab=12, nazwa_pliku="jan_kowalski.json", ascii=True)
```

Sprawdźmy co zostało zapisane do pliku:

```python
with open("jan_kowalski.json") as p:
    print(p.read())
```

Załóżmy, że konfigurację w naszym programie przechowujemy w pliku w formacie [TOML](https://docs.python.org/3/library/tomllib.html). Po wczytaniu danych z pliku mogą być one przechowywane jako napis:

```python
konfiguracja_toml = 'spacje = true\nrozmiar_tab = 4\nnazwa_pliku = "dane_osobowe.json"\nsortuj_klucze = true\nascii = false'
```

```python
print(konfiguracja_toml)
```

Nazwy opcji konfiguracji są takie same jak nazwy parametrów funkcji `zapisz_wyniki`.

Moduł [`tomllib`](https://docs.python.org/3/library/tomllib.html) umożliwia łatwą zamianę takich danych na odpowiedni słownik:

```python
import tomllib
```

```python
konfiguracja = tomllib.loads(konfiguracja_toml)
```

```python
konfiguracja
```

Możemy teraz wywołać funkcję `zapisz_wyniki` w taki sposób:

```python
zapisz_wyniki(
    dane,
    spacje=konfiguracja["spacje"],
    rozmiar_tab=konfiguracja["rozmiar_tab"],
    nazwa_pliku=konfiguracja["nazwa_pliku"],
    sortuj_klucze=konfiguracja["sortuj_klucze"],
    ascii=konfiguracja["ascii"],
)
```

Ale przy użyciu opisanego mechanizmu będzie to wyglądać tak:

```python
zapisz_wyniki(dane, **konfiguracja)
```

## Funkcja przyjmująca dowolne argumenty

Do przykładów poniżej przydadzą się zdefiniowane wcześniej funkcje `srednia` i `srednia_arytmetyczna`. Powtarzamy tu ich definicje:

```python
def srednia(dane, /, wagi=None, *, dokladnosc=None):
    if wagi is None:
        wagi = [1] * len(dane)
    wynik = sum(d * w for d, w in zip(dane, wagi)) / sum(wagi)
    if dokladnosc is not None:
        wynik = round(wynik, dokladnosc)
    return wynik
```

```python
def srednia_arytmetyczna(x, *xs, wyswietl=False):
    if wyswietl:
        print(f"x = {x}, xs = {xs}")
    return (x + sum(xs)) / (1 + len(xs))
```

Poniżej podany jest przykład funkcji, która używa wszystkich czterech opisanych powyżej mechanizmów. Funkcja może przyjąć zarówno dowolną liczbę argumentów pozycyjnych jak i dowolne argumenty nazwane. Funkcja wywołuje inną funkcję i przekazuje jej otrzymaną krotkę jako osobne argumenty pozycyjne, oraz otrzymany słownik jako osobne argumenty nazwane.

Jest to przykład tzw. funkcji *opakowującej* inną funkcję. Funkcja dostaje jako pierwszy argument inną funkcję a jako kolejne argumenty dostaje argumenty do wywołania tej funkcji (zarówno pozycyjne jak i nazwane). Funkcja po prostu wywołuje tę funkcję używając zadanych parametrów, a oprócz tego wyświetla komunikat o wywołaniu tej funkcji, jej argumentach i wyniku.

Nazwy `args` i `kwargs` są powszechnie używanymi nazwami dla tego typu parametrów.

```python
def wywolaj_funkcje(f, *args, **kwargs):
    print(f"wywołanie funkcji '{f.__name__}'")
    print(f"argumenty pozycyjne: {args}")
    print(f"argumenty nazwane: {kwargs}")
    wynik = f(*args, **kwargs)
    print(f"wynik = {wynik}")
    return wynik
```

```python
from math import sin, sqrt, pi
```

```python
wywolaj_funkcje(sin, pi / 2)
```

```python
wywolaj_funkcje(srednia_arytmetyczna, 2, 3, 4, 5)
```

```python
wywolaj_funkcje(srednia, [2, 3, 4], wagi=[1, 2, 4], dokladnosc=3)
```

Funkcja opakowująca przyjmie poniższe argumenty, ale ich zgodność z parametrami sprawdzi dopiero wywoływana funkcja `srednia`. Ostatnia liczba ma oznaczać dokładność, a została podana pozycyjnie. Funkcja `srednia` wymaga przekazania jej jako argumentu nazwanego `dokladnosc`, więc zgłosi wyjątek [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

```python
wywolaj_funkcje(srednia, [2, 3, 4], [3, 4, 5], 2)
```

### Rozpakowanie w listach i słownikach

Podobny zapis pozwala wstawić elementy sekwencji do nowej listy i wpisy słownika do nowego słownika. Nie jest ograniczony do wywołań funkcji.

```python
a = [1, 2, 3]
b = [0, 0, *a, 0, 0, *a, *range(10)]
```

```python
b
```

```python
d = {"a": 1, "b": 2, "c": 3}
e = {"d": 4, "e": 5}
```

```python
{**d, **e, "f": 6}
```

```python
d
```

```python
e
```

W literałach słowników późniejszy wpis o tym samym kluczu zastępuje wcześniejszy. W wywołaniu funkcji dwukrotne podanie wartości tego samego parametru jest natomiast błędem.

```python
{**{"a": 1}, **{"a": 2}}
```
