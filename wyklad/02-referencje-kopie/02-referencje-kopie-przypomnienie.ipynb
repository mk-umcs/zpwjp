# Referencje i kopie — typy wbudowane

## Obiekty mutowalne i niemutowalne

Obiekt **mutowalny** (ang. *mutable*) pozwala zmieniać swój stan po utworzeniu. Na przykład można zmienić element listy albo dodać wpis do słownika.

Obiektu **niemutowalnego** (ang. *immutable*) nie można w ten sposób zmienić. Liczby, napisy i krotki są niemutowalne. Operacja dająca inną wartość może powiązać zmienną z innym obiektem zamiast zmieniać dotychczasowy.

W obu przypadkach zmienne wskazują obiekty. Przypisanie nie ma osobnych reguł dla „typów prostych” i „złożonych”. Różne skutki wynikają z mutowalności i rodzaju operacji.

## Przypisanie nie tworzy kopii

Jak działa instrukcja podstawienia w przypadku list?

```python
lista1 = [1, 2, 3]
```

```python
lista2 = lista1
```

Czy powstała kopia? Zmieńmy element przez drugą zmienną:

```python
lista2[1] = 7
```

```python
lista1[1]
```

Zmiana jest widoczna również przez `lista1`: obie zmienne wskazują tę samą listę. Mówimy, że są **referencjami** (ang. *references*) do tego samego obiektu.

Operator `is` sprawdza, czy chodzi o ten sam obiekt. Operator `==` sprawdza równość według reguł danego typu.

```python
lista1 is lista2
```

### Liczby i ponowne przypisanie

Liczba całkowita jest niemutowalna. Instrukcja `n += 5` w poniższym przykładzie wiąże nazwę `n` z obiektem reprezentującym wynik dodawania, zamiast zmieniać liczbę wskazywaną wcześniej.

```python
n = 10
```

```python
poprzednia = n
```

```python
n += 5
```

```python
n
```

```python
poprzednia
```

Do porównywania wartości liczb i napisów używamy operatora `==`, a nie `is`.

## Zmiana listy a przypisanie w funkcji

Argument funkcji zostaje powiązany z lokalnym parametrem. Przypisanie nowej listy do parametru nie zmienia powiązania zmiennej, której użyliśmy przy wywołaniu.

```python
def przypisz_pusta(lista):
    lista = []
    return lista
```

```python
oceny = [3, 4, 5]
```

```python
nowa = przypisz_pusta(oceny)
```

```python
oceny
```

```python
nowa
```

Modyfikacja wspólnej listy jest natomiast widoczna również poza funkcją:

```python
def wyczysc(lista):
    lista.clear()
    return lista
```

```python
wynik = wyczysc(oceny)
```

```python
oceny
```

```python
wynik is oceny
```

Samo zwrócenie listy z funkcji również nie tworzy jej kopii.

## Płytka kopia listy

Metoda [`list.copy()`](https://docs.python.org/3/library/stdtypes.html#list.copy) tworzy nową listę, której elementy wskazują te same obiekty co elementy oryginału. To **płytka kopia** (ang. *shallow copy*).

```python
pierwsza = [1, 2, 3]
```

```python
druga = pierwsza.copy()
```

```python
druga is pierwsza
```

```python
druga[0] = 100
```

```python
pierwsza
```

```python
druga
```

Podobnie można utworzyć kopię listy za pomocą wycinka obejmującego całą listę:

```python
kopia_wycinek = pierwsza[:]
```

```python
kopia_wycinek is pierwsza
```

```python
kopia_wycinek == pierwsza
```

### Listy zagnieżdżone

Płytka kopia listy nie kopiuje list znajdujących się w jej elementach.

```python
lista = [1, [2, 3]]
```

```python
kopia = lista.copy()
```

```python
kopia[1] is lista[1]
```

```python
kopia[0] = 100
```

```python
kopia[1].append(4)
```

```python
lista
```

```python
kopia
```

Podmiana pierwszego elementu zmieniła tylko kopię. Dopisanie do listy zagnieżdżonej zmieniło jednak wspólny obiekt. Wycinek `lista[:]` dałby taką samą płytką kopię.

## Głęboka kopia listy list

Jeżeli chcemy niezależnie zmieniać także listy zagnieżdżone, potrzebujemy kopii obu poziomów. Taką kopię nazywamy **głęboką** (ang. *deep copy*).

Dla listy, której elementami są listy liczb, możemy wykonać ją ręcznie:

```python
wiersze = [[1, 2], [3, 4]]
```

```python
kopie_wierszy = []
for wiersz in wiersze:
    kopie_wierszy.append(wiersz.copy())
```

```python
kopie_wierszy is wiersze
```

```python
kopie_wierszy[0] is wiersze[0]
```

```python
kopie_wierszy[0][0] = 100
```

```python
wiersze
```

```python
kopie_wierszy
```

Kopiujemy listę zewnętrzną i każdy jej wiersz. Liczby mogą pozostać wspólne, bo są niemutowalne. Dla większej liczby mutowalnych poziomów trzeba byłoby skopiować także te poziomy. Ta pętla jest rozwiązaniem dla opisanej struktury, a nie uniwersalnym sposobem kopiowania dowolnych danych.

## Słownik z listami jako wartościami

Metoda [`dict.copy()`](https://docs.python.org/3/library/stdtypes.html#dict.copy) również tworzy płytką kopię. Powstaje nowy słownik, ale przechowywane listy pozostają wspólne.

```python
oceny = {"Jan": [3, 4], "Anna": [5]}
```

```python
kopia_ocen = oceny.copy()
```

```python
kopia_ocen is oceny
```

```python
kopia_ocen["Jan"] is oceny["Jan"]
```

```python
kopia_ocen["Jan"].append(5)
```

```python
oceny["Jan"]
```

## Niemutowalna krotka z mutowalnym elementem

Krotka jest niemutowalna: nie można podmienić jej elementu. Może jednak zawierać referencję do listy, którą nadal można zmieniać.

```python
para = ("oceny", [3, 4])
```

```python
para[1].append(5)
```

```python
para
```

Nie zmieniliśmy tego, na jakie obiekty wskazują elementy krotki. Zmieniliśmy zawartość listy wskazywanej przez drugi element. Próba `para[1] = [1, 2]` zakończyłaby się natomiast błędem [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError).

## Powielanie listy operatorem `*`

Powielanie listy powiela referencje do jej elementów, a nie same obiekty.

```python
wiersze = [[0, 0]] * 3
```

```python
wiersze
```

```python
wiersze[0] is wiersze[1]
```

```python
wiersze[0][0] = 9
```

```python
wiersze
```

Każdy z trzech elementów wskazuje ten sam wiersz. Osobne wiersze można utworzyć na przykład wyrażeniem listowym:

```python
niezalezne_wiersze = [[0, 0] for _ in range(3)]
```

```python
niezalezne_wiersze[0] is niezalezne_wiersze[1]
```

```python
niezalezne_wiersze[0][0] = 9
```

```python
niezalezne_wiersze
```
