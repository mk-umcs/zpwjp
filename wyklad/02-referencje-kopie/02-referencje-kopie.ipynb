## Referencje

Poniższy ciąg instrukcji:

```python
a = [1, 2, 3]
```

```python
b = a
```

```python
b[1] = 7
```

```python
a[1]
```

Pokazuje, że instrukcja podstawienia nie tworzy kopii &mdash; zmienne `a` i `b` wskazują na tę samą listę. Są **referencją** do tej samej listy. Modyfikacja listy wskazywanej przez `b` jest widoczna również jeżeli użyjemy zmiennej `a`. Jak to dokładnie działa można sprawdzić na stronie [Python Tutor](https://pythontutor.com/render.html#code=a%20%3D%20%5B1,%202,%203%5D%0Ab%20%3D%20a%0Ab%5B1%5D%20%3D%207%0A%0Aprint%28a%5B1%5D%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&origin=opt-frontend.js&py=3&rawInputLstJSON=%5B%5D&textReferences=false).

Ten sam mechanizm ma zastosowanie do obiektów:

```python
class Osoba:
    
    def __init__(self, imie, nazwisko, rok_urodzenia):
        self.imie = imie
        self.nazwisko = nazwisko
        self.rok_urodzenia = rok_urodzenia
```

```python
osoba_1 = Osoba("Jan", "Kowalski", 2000)
```

```python
osoba_2 = osoba_1
```

```python
osoba_2.rok_urodzenia = 2010
```

```python
osoba_1.rok_urodzenia
```

Zmienna `osoba_2` jest teraz referencją do tego samego obiektu co zmienna `osoba_1`.

To samo dotyczy obiektów, które same są polami innych obiektów. Zdefiniujmy klasę do reprezentowania daty:

```python
class Data:

    def __init__(self, rok, miesiac, dzien):
        self.rok = rok
        self.miesiac = miesiac
        self.dzien = dzien
        
    def __repr__(self):
        return f"Data({self.rok!r}, {self.miesiac!r}, {self.dzien!r})"
```

a następnie klasę `Osoba` z polem `data_urodzenia`:

```python
class Osoba:
    
    def __init__(self, imie, nazwisko, data_urodzenia):
        self.imie = imie
        self.nazwisko = nazwisko
        self.data_urodzenia = data_urodzenia
```

Tworzymy obiekt reprezentujący dzień 10 maja 2010 roku:

```python
dzien = Data(2010, 5, 10)
```

Tworzymy dwa obiekty reprezentujące osoby urodzone tego dnia:

```python
jan = Osoba("Jan", "Kowalski", dzien)
```

```python
anna = Osoba("Anna", "Nowak", dzien)
```

Zmieniamy dzień urodzenia jednej z osób:

```python
anna.data_urodzenia.dzien += 10
```

Sprawdzamy datę urodzenia drugiej osoby:

```python
jan.data_urodzenia
```

Widzimy, że data została zmieniona. Spowodowane jest to tym, że osoby dzielą wspólny obiekt `Data`. Sprawdź to na stronie [Python Tutor](https://pythontutor.com/visualize.html#code=class%20Data%3A%0A%0A%20%20%20%20def%20__init__%28self,%20rok,%20miesiac,%20dzien%29%3A%0A%20%20%20%20%20%20%20%20self.rok%20%3D%20rok%0A%20%20%20%20%20%20%20%20self.miesiac%20%3D%20miesiac%0A%20%20%20%20%20%20%20%20self.dzien%20%3D%20dzien%0A%20%20%20%20%20%20%20%20%0A%20%20%20%20def%20__repr__%28self%29%3A%0A%20%20%20%20%20%20%20%20return%20f%22Data%28%7Bself.rok!r%7D,%20%7Bself.miesiac!r%7D,%20%7Bself.dzien!r%7D%29%22%0A%20%20%20%20%20%20%20%20%0Aclass%20Osoba%3A%0A%20%20%20%20%0A%20%20%20%20def%20__init__%28self,%20imie,%20nazwisko,%20data_urodzenia%29%3A%0A%20%20%20%20%20%20%20%20self.imie%20%3D%20imie%0A%20%20%20%20%20%20%20%20self.nazwisko%20%3D%20nazwisko%0A%20%20%20%20%20%20%20%20self.data_urodzenia%20%3D%20data_urodzenia%0A%20%20%20%20%20%20%20%20%0Adzien%20%3D%20Data%282010,%205,%2010%29%0Ajan%20%3D%20Osoba%28%22Jan%22,%20%22Kowalski%22,%20dzien%29%0Aanna%20%3D%20Osoba%28%22Anna%22,%20%22Nowak%22,%20dzien%29%0A%0Aanna.data_urodzenia.dzien%20%2B%3D%2010&cumulative=false&heapPrimitives=nevernest&mode=edit&origin=opt-frontend.js&py=3&rawInputLstJSON=%5B%5D&textReferences=false).

## Kopie

Jeżeli chcemy, żeby obiekty były kopiowane, to możemy zrobić to np. tak (podczas tworzenia obiektu klasy `Osoba` tworzony jest też zawsze nowy obiekt `Data`):

```python
class Osoba:
    
    def __init__(self, imie, nazwisko, data_urodzenia):
        self.imie = imie
        self.nazwisko = nazwisko
        self.data_urodzenia = Data(data_urodzenia.rok, data_urodzenia.miesiac, data_urodzenia.dzien)
```

```python
jan = Osoba("Jan", "Kowalski", dzien)
```

```python
anna = Osoba("Anna", "Nowak", dzien)
```

```python
anna.data_urodzenia.miesiac = 2
```

```python
jan.data_urodzenia.miesiac = 7
```

```python
anna.data_urodzenia
```

```python
jan.data_urodzenia
```

```python
dzien
```

Widać, że tym razem mamy trzy osobne obiekty typu `Data`.

### Moduł [`copy`](https://docs.python.org/3/library/copy.html)

Do kopiowania obiektów możemy też użyć funkcji z modułu [`copy`](https://docs.python.org/3/library/copy.html):
* płytka kopia: [copy.copy](https://docs.python.org/3/library/copy.html#copy.copy)
* głęboka kopia: [copy.deepcopy](https://docs.python.org/3/library/copy.html#copy.deepcopy)

Poniższe przykłady pokazują dokładnie na czym polega płytka i głęboka kopia:

```python
import copy
```

```python
osoba = Osoba("Jan", "Kowalski", dzien)
```

Tutaj nic nie jest kopiowane, mamy dwie zmienne wskazujące na ten sam obiekt typu `Osoba`:

```python
referencja = osoba
```

Stworzenie płytkiej kopii:

```python
plytka_kopia = copy.copy(osoba)
```

Skopiowane są wszystkie pola, ale jeżeli pole jest referencją do obiektu (również listy, krotki), to skopiowana będzie referencja. W tym przypadku pole `data_urodzenia` typu `Data` będzie wspólne.

```python
gleboka_kopia = copy.deepcopy(osoba)
```

W tym przypadku wszystko będzie skopiowane. Obiekt wskazywany przez `gleboka_kopia` nie ma żadnych danych wspólnych z obiektem `osoba`.

Podobnie działa to np. dla list (oraz innych typów złożonych takich jak słowniki czy krotki).
Poniżej tworzymy listę, której ostatnim elementem jest również lista.

```python
lista = [1, 2, [3, 4]]
```

Tworzymy kopię listy:

```python
kopia_listy = copy.copy(lista)
```

```python
kopia_listy
```

Lista została skopiowana, ale ostatni element (lista `[3, 4]`) jest wspólny:

```python
kopia_listy[0] = -1
kopia_listy[2][0] = -3
```

```python
kopia_listy
```

```python
lista
```

Tworzymy głęboką kopię.

```python
gleboka_kopia_listy = copy.deepcopy(lista)
```

Skopiowana lista nie ma żadnych wspólnych danych z listą.

```python
gleboka_kopia_listy[2][0] = 1000
```

```python
gleboka_kopia_listy
```

```python
lista
```
