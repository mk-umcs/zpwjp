# Dziedziczenie

## Dziedziczenie

Mechanizm dziedziczenia umożliwia tworzenie hierarchii klas. Ułatwia tworzenie klas &bdquo;bardziej wyspecjalizowanych&rdquo;. Np. po stworzeniu klasy `Osoba` (przechowującej np. informacje o imieniu, nazwisku, numerze PESEL, umożliwiającej np. zmianę imienia, obliczenie wieku) możemy użyć jej do stworzenia bardziej wyspecjalizowanej klasy `Pracownik`. Pracownik jest osobą więc obiekty tej klasy przechowują te same informacje oraz umożliwiają wykonanie tych samych operacji co obiekty klasy `Osoba`. Możliwe jest dodanie dodatkowych danych i operacji (lub zmiana sposobu działania operacji z klasy `Osoba`). Podobnie możemy stworzyć np. klasę `Student` (student jest osobą) lub `Lekarz` (lekarz jest zarówno osobą jak i pracownikiem) czy `Nauczyciel` (też osoba i pracownik).

Efektem może być poniższa hierarchia:

```text
Osoba
├── Student
└── Pracownik
    ├── Lekarz
    └── Nauczyciel
```

Mówimy, że klasa `Student` **dziedziczy** po klasie `Osoba`. Klasa `Osoba` jest **klasą bazową**/**nadklasą** (ang. superclass) klasy `Student`. Klasa `Student` jest **klasą pochodną**/**podklasą** (ang. subclass) klasy `Osoba`.

Terminy &bdquo;nadklasa&rdquo; i &bdquo;podklasa&rdquo; mogą kojarzyć się z matematycznymi terminami &bdquo;nadzbiór&rdquo; i &bdquo;podzbiór&rdquo;. Zbiór osób jest nadzbiorem zbioru studentów, a zbiór studentów jest podzbiorem zbioru osób.

### Przykład

Zaczniemy od zdefiniowania klasy `Osoba` z właściwościami do odczytu `imie`, `nazwisko`, właściwością do odczytu i zapisu `wiek` oraz z metodą `pelnoletni`.

````python
class Osoba:
    def __init__(self, imie, nazwisko, wiek):
        if not isinstance(imie, str):
            raise TypeError("Imię musi być typu str")
        if not isinstance(nazwisko, str):
            raise TypeError("Nazwisko musi być typu str")
        self._imie = imie
        self._nazwisko = nazwisko
        self._wiek = wiek

    def __repr__(self):
        return f"Osoba({self.imie!r}, {self.nazwisko!r}, {self.wiek!r})"

    @property
    def imie(self):
        return self._imie

    @property
    def nazwisko(self):
        return self._nazwisko

    @property
    def wiek(self):
        return self._wiek

    @wiek.setter
    def wiek(self, w):
        self._wiek = w

    def pelnoletni(self):
        return self._wiek >= 18
````

````python
osoba = Osoba("Jan", "Kowalski", 17)
````

````python
osoba
````

````python
osoba.imie
````

````python
osoba.nazwisko
````

````python
osoba.wiek
````

````python
osoba.pelnoletni()
````

````python
osoba.wiek = 18
````

````python
osoba.wiek
````

````python
osoba.pelnoletni()
````

Jeżeli chcemy stworzyć klasę dziedziczącą to nazwę klasy bazowej podajemy w nawiasach po nazwie klasy pochodnej. Np. poniższa definicja:

````python
class Pracownik(Osoba):
    pass
````

oznacza stworzenie klasy `Pracownik`, której obiekty zachowują się prawie dokładnie tak samo jak obiekty klasy `Osoba`. Wszystkie składowe (pola, metody, właściwości) zostały odziedziczone.

````python
pracownik = Pracownik("Jan", "Kowalski", 30)
````

````python
pracownik.imie
````

````python
pracownik.pelnoletni()
````

Nazwy rozpoczynające się podkreśleniem są niepubliczne z konwencji. Technicznie można je odczytać, ale kod korzystający z klasy nie powinien od nich zależeć:

````python
pracownik._imie
````

Metoda `__repr__` również została odziedziczona:

````python
pracownik
````

Możemy **przesłonić**/**nadpisać**/**przedefiniować** (ang. *override*) odziedziczone metody oraz dodawać nowe składowe. W poniższej definicji klasy `Pracownik` przesłonięta jest metoda `__repr__` oraz dodane są pola `_pensja`, właściwość `pensja` oraz metoda `roczne_zarobki`.

````python
class Pracownik(Osoba):
    def __init__(self, imie, nazwisko, wiek, pensja):
        super().__init__(imie, nazwisko, wiek)
        self._pensja = pensja

    def __repr__(self):
        return f"Pracownik({self.imie!r}, {self.nazwisko!r}, {self.wiek!r}, {self.pensja!r})"

    @property
    def pensja(self):
        return self._pensja

    def roczne_zarobki(self):
        return self._pensja * 12
````

Konstrukcja [`super()`](https://docs.python.org/3/library/functions.html#super) (użyta powyżej w konstruktorze) umożliwia odwołanie do składowych klasy bazowej. Wywołanie metody `__init__` klasy bazowej w metodzie `__init__` klasy pochodnej jest jedną z sytuacji, w których powinno się jawnie wywołać metodę specjalną. Zazwyczaj metoda `__init__` wywoływana jest podczas tworzenia obiektu, ale w tym przypadku obiekt został już utworzony.

````python
pracownik = Pracownik("Jan", "Kowalski", 30, 3000)
````

Użycie nadpisanej metody `__repr__`:

````python
pracownik
````

````python
pracownik.imie
````

````python
pracownik.pelnoletni()
````

Użycie dodanych składowych:

````python
pracownik.pensja
````

````python
pracownik.roczne_zarobki()
````

### Funkcja [`isinstance`](https://docs.python.org/3/library/functions.html#isinstance)

Typ obiektu możemy sprawdzić za pomocą funkcji [`type`](https://docs.python.org/3/library/functions.html#type):

````python
type(pracownik)
````

````python
type(pracownik) is Pracownik
````

Pracownik jest osobą (wszędzie tam gdzie możemy użyć obiektu klasy `Osoba` powinniśmy móc użyć obiektu klasy `Pracownik`). Niestety nie możemy tego sprawdzić za pomocą funkcji [`type`](https://docs.python.org/3/library/functions.html#type):

````python
type(pracownik) is Osoba
````

Do sprawdzenia czy typem obiektu jest zadana klasa lub jedna z jej klas pochodnych służy funkcja [`isinstance`](https://docs.python.org/3/library/functions.html#isinstance):

````python
isinstance(pracownik, Pracownik)
````

````python
isinstance(pracownik, Osoba)
````

Jeżeli chcemy np. sprawdzić czy argument funkcji jest odpowiedniego typu, to zazwyczaj lepszym pomysłem będzie użycie [`isinstance`](https://docs.python.org/3/library/functions.html#isinstance) zamiast sprawdzenia typu funkcją [`type`](https://docs.python.org/3/library/functions.html#type). 

Rozważmy poniższą funkcję:

````python
def info(osoba):
    if type(osoba) is not Osoba:
        raise TypeError("Argument złego typu")
    print(f"imię: {osoba.imie}")
    print(f"nazwisko: {osoba.nazwisko}")
    print(f"wiek: {osoba.wiek}")
````

````python
info(osoba)
````

Funkcja `info` odrzuci jednak obiekt `pracownik`, mimo że jest on osobą. Sprawdza dokładny typ, więc dla podklasy `Pracownik` zgłosi [`TypeError`](https://docs.python.org/3/library/exceptions.html#TypeError):

````python
info(pracownik)
````

Sprawdzenie typu powinno raczej wyglądać tak:

````python
def info(osoba):
    if not isinstance(osoba, Osoba):
        raise TypeError("Argument złego typu")
    print(f"imię: {osoba.imie}")
    print(f"nazwisko: {osoba.nazwisko}")
    print(f"wiek: {osoba.wiek}")
````

````python
info(osoba)
````

````python
info(pracownik)
````

Funkcja [`isinstance`](https://docs.python.org/3/library/functions.html#isinstance) może być użyta również do obiektów typów wbudowanych:

````python
isinstance([1, 2, 3], list)
````

````python
isinstance(5, int)
````

````python
isinstance({}, dict)
````

Jako drugiego argumentu można również użyć krotki typów:

````python
liczba = 3
````

````python
isinstance(liczba, (int, float))
````

### Dziedziczenie a hermetyzacja

W klasie pochodnej nie powinno się korzystać ze składowych niepublicznych klasy bazowej.

Niech klasa `PracownikZPremia` umożliwia przyznawanie pracownikom premii. Pracownikom można przyznawać premie, a ich roczne zarobki, to suma miesięcznych pensji i przyznanych premii. Zwróć uwagę na definicję metody `roczne_zarobki` (zostało tam użyte niepubliczne pole `_pensja` z klasy `Pracownik`):

Dla wygody powtarzamy obie klasy potrzebne do przykładu premii. Ich publiczny interfejs pozostaje taki sam:

````python
class Osoba:
    def __init__(self, imie, nazwisko, wiek):
        if not isinstance(imie, str):
            raise TypeError("Imię musi być typu str")
        if not isinstance(nazwisko, str):
            raise TypeError("Nazwisko musi być typu str")
        self._imie = imie
        self._nazwisko = nazwisko
        self._wiek = wiek

    def __repr__(self):
        return f"Osoba({self.imie!r}, {self.nazwisko!r}, {self.wiek!r})"

    @property
    def imie(self):
        return self._imie

    @property
    def nazwisko(self):
        return self._nazwisko

    @property
    def wiek(self):
        return self._wiek

    @wiek.setter
    def wiek(self, w):
        self._wiek = w

    def pelnoletni(self):
        return self._wiek >= 18
````

````python
class Pracownik(Osoba):
    def __init__(self, imie, nazwisko, wiek, pensja):
        super().__init__(imie, nazwisko, wiek)
        self._pensja = pensja

    def __repr__(self):
        return f"Pracownik({self.imie!r}, {self.nazwisko!r}, {self.wiek!r}, {self.pensja!r})"

    @property
    def pensja(self):
        return self._pensja

    def roczne_zarobki(self):
        return self._pensja * 12
````

````python
class PracownikZPremia(Pracownik):
    def __init__(self, imie, nazwisko, wiek, pensja):
        super().__init__(imie, nazwisko, wiek, pensja)
        self._premia = 0

    def przyznaj_premie(self, premia):
        self._premia += premia

    def roczne_zarobki(self):
        return self._pensja * 12 + self._premia
````

````python
pracownik = PracownikZPremia("Jan", "Kowalski", 30, 3000)
````

````python
pracownik.przyznaj_premie(4000)
````

````python
pracownik.roczne_zarobki()
````

Załóżmy, że chcemy teraz zmodyfikować implementację klasy `Pracownik`. Zamiast przechowywać pensję chcemy przechowywać sumę dwunastu pensji, a pensję obliczać w razie potrzeby. Modyfikacja taka nie powinna powodować żadnych zmian w kodzie używającym tej klasy:

````python
class Pracownik(Osoba):
    def __init__(self, imie, nazwisko, wiek, pensja):
        super().__init__(imie, nazwisko, wiek)
        self._roczne_zarobki = pensja * 12

    def __repr__(self):
        return f"Pracownik({self.imie!r}, {self.nazwisko!r}, {self.wiek!r}, {self.pensja!r})"

    @property
    def pensja(self):
        return self._roczne_zarobki / 12

    def roczne_zarobki(self):
        return self._roczne_zarobki
````

Niestety teraz przestanie działać metoda `roczne_zarobki` z klasy `PracownikZPremia`:

````python
class PracownikZPremia(Pracownik):
    def __init__(self, imie, nazwisko, wiek, pensja):
        super().__init__(imie, nazwisko, wiek, pensja)
        self._premia = 0

    def przyznaj_premie(self, premia):
        self._premia += premia

    def roczne_zarobki(self):
        return self._pensja * 12 + self._premia
````

````python
pracownik = PracownikZPremia("Jan", "Kowalski", 30, 3000)
````

Wywołanie `roczne_zarobki` próbuje nadal odczytać pole `_pensja`, którego nowa wersja klasy bazowej już nie tworzy. Otrzymamy więc [`AttributeError`](https://docs.python.org/3/library/exceptions.html#AttributeError):

````python
pracownik.roczne_zarobki()
````

Jest to spowodowane użyciem składowej niepublicznej. Żeby uniknąć tego problemu możemy w metodzie `roczne_zarobki` klasy `PracownikZPremia` użyć metody `roczne_zarobki` z jej klasy bazowej `Pracownik`:

````python
class PracownikZPremia(Pracownik):
    def __init__(self, imie, nazwisko, wiek, pensja):
        super().__init__(imie, nazwisko, wiek, pensja)
        self._premia = 0

    def przyznaj_premie(self, premia):
        self._premia += premia

    def roczne_zarobki(self):
        return super().roczne_zarobki() + self._premia
````

Podobnie jak w konstruktorze, tutaj również użyliśmy konstrukcji [`super()`](https://docs.python.org/3/library/functions.html#super).

````python
pracownik = PracownikZPremia("Jan", "Kowalski", 30, 3000)
````

````python
pracownik.roczne_zarobki()
````

### Klasa [`object`](https://docs.python.org/3/library/functions.html#object)

Wszystkie klasy dziedziczą bezpośrednio lub pośrednio po klasie [`object`](https://docs.python.org/3/library/functions.html#object):

````python
isinstance(pracownik, object)
````

````python
isinstance(5, object)
````

Jeżeli nie podamy klasy bazowej w definicji klasy, to domyślnie jest to klasa [`object`](https://docs.python.org/3/library/functions.html#object), czyli np. deklaracja

````python
class A:
    pass
````

jest równoważna tej:

````python
class A(object):
    pass
````

Domyślne implementacje np. metod `__str__`, `__repr__` czy `__eq__` lub `__lt__` są w klasach odziedziczone właśnie po klasie [`object`](https://docs.python.org/3/library/functions.html#object).
